```python
# Convert Order Date into datetime format
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
```

```python
# Convert Ship Date into datetime format
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)
```

```python
# Verify that date columns have been converted successfully
df.info()
```

```python
#Extract year from Order Date 
df["Year"] = df["Order Date"].dt.year
```

```python
# Extract month name from Order Date
df["Month"] = df["Order Date"].dt.month_name()
```

```python
# Extract month number for correct sorting
df["Month Number"] = df["Order Date"].dt.month
```

```python
# Extract quarter from Order Date
df["Quarter"] = df["Order Date"].dt.quarter
```

```python
# Display first five rows with newly created columns
df.head()
```

```python
# Calculate total sales for each month
monthly_sales = df.groupby(["Month Number", "Month"])["Sales"].sum().reset_index()

# Sort data by month number
monthly_sales = monthly_sales.sort_values("Month Number")

# Display the result
monthly_sales
```

```python
#Plot monthly sales trend
# Set figure size
plt.figure(figsize=(12,6))

# Create line chart
sns.lineplot(data=monthly_sales,
             x="Month",
             y="Sales",
             marker="o",
             color="blue")

# Add title
plt.title("Monthly Sales Trend", fontsize=16)

# Add labels
plt.xlabel("Month")
plt.ylabel("Total Sales")

# Rotate month names
plt.xticks(rotation=45)

# Display chart
plt.show()
```

## Observation

# Sales increased during the last quarter of the year.
# November and December recorded higher sales.
# February had comparatively lower sales.

```python
# Calculate total sales for each quarter
quarterly_sales = df.groupby("Quarter")["Sales"].sum().reset_index()

# Display the result
quarterly_sales
```

```python
#PLot Quarterly Sales Trend
#Set figure size
plt.figure(figsize=(8,5))

# Create line chart
sns.lineplot(
    data=quarterly_sales,
    x="Quarter",
    y="Sales",
    marker="o",
    color="green"
)

# Add title and labels
plt.title("Quarterly Sales Trend", fontsize=14)
plt.xlabel("Quarter")
plt.ylabel("Total Sales")

# Show the graph
plt.show()
```

## Observations

#  Quarter 4 recorded the highest sales.
#  Quarter 1 recorded the lowest sales.
#  Sales increased/decreased from Quarter 1to Quarter 4
# This indicates stronger customer purchasing activity towards the end of the year

```python
# Calculate total sales for each product
top_products = df.groupby("Product Name")["Sales"].sum().reset_index()

# Sort products by sales in descending order
top_products = top_products.sort_values(by="Sales", ascending=False)

# Select top 10 products
top_products = top_products.head(10)

# Display the result
top_products
```

```python
#Plot Top 10 Best Selling Products
# Set figure size
plt.figure(figsize=(14,6))

# Create horizontal bar chart
sns.barplot(
    data=top_products,
    x="Sales",
    y="Product Name",
    hue="Product Name",
    palette="coolwarm",
    legend=False
)

# Add chart title
plt.title("Top 10 Best-Selling Products", fontsize=16, fontweight="bold")

# Add axis labels
plt.xlabel("Total Sales")
plt.ylabel("Product Name")

# Improve layout
plt.tight_layout()

# Display the chart
plt.show()
```

## Observations

# The **Canon imageCLASS 2200 Advanced Copier** generated the highest sales among all products.
# The top 10 best-selling products contribute significantly to the company's overall revenue.
# High-performing products should be kept in stock to avoid inventory shortages.
# The company can increase sales further by promoting these top-selling products through marketing campaigns.

```python
# Group the dataset by Category and calculate total sales
category_sales = df.groupby("Category")["Sales"].sum().reset_index()

# Sort categories based on total sales in descending order
category_sales = category_sales.sort_values(by="Sales", ascending=False)

# Display the result
category_sales
```

```python
#Plot Revenue by Product Category
# Set figure size
plt.figure(figsize=(8,5))

# Create bar chart
sns.barplot(
    data=category_sales,
    x="Category",
    y="Sales",
    hue="Category",
    palette="Set2",
    legend=False
)

# Add chart title
plt.title("Revenue by Product Category", fontsize=16, fontweight="bold")

# Add axis labels
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")

# Improve layout
plt.tight_layout()

# Display chart
plt.show()
```

## Observations

# The Technology category generated the highest revenue.
# The Office Supplies category generated the lowest revenue.
# Technology/Furniture/Office Supplies contributed significantly to total sales.
# The company should focus on high-performing categories while improving the performance of low-selling categories.

```python
# Calculate correlation between numerical columns
correlation_matrix = df.corr(numeric_only=True)

# Display correlation matrix
correlation_matrix
```

```python
# Plot Correlation Heatmap
# Set figure size
plt.figure(figsize=(8,6))

# Create heatmap
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5,
    fmt=".2f"
)

# Add title
plt.title("Correlation Heatmap", fontsize=16, fontweight="bold")

# Display heatmap
plt.show()
```

## Observations

# The strongest positive correlation (**0.97**) exists between **Month Number** and **Quarter**, as both are time-related variables.
# **Sales** shows very weak correlation with the other numerical variables, indicating that sales are not directly influenced by these variables.
# **Postal Code** and **Row ID** also have negligible correlation with Sales.
# Overall, there is no strong positive or negative correlation between Sales and the available numerical features.

```python
# Group the dataset by Region and calculate total sales
region_sales = df.groupby("Region")["Sales"].sum().reset_index()

# Sort regions based on total sales
region_sales = region_sales.sort_values(by="Sales", ascending=False)

# Display the result
region_sales
```

```python
#Total Sales by region
#Set figure size
plt.figure(figsize=(8,5))

# Create bar chart
sns.barplot(
    data=region_sales,
    x="Region",
    y="Sales",
    hue="Region",
    palette="Set3",
    legend=False
)

# Add title
plt.title("Total Sales by Region", fontsize=16, fontweight="bold")

# Add axis labels
plt.xlabel("Region")
plt.ylabel("Total Sales")

# Improve layout
plt.tight_layout()

# Display chart
plt.show()
```

## Observations

#The West region generated the highest total sales.
# The South region generated the lowest total sales.
# Sales vary across different regions, indicating differences in customer demand.
# The company can focus on improving sales in lower-performing regions while maintaining strong performance in high-selling regions.
