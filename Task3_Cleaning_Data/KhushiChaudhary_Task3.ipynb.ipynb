```python
#Step 1: Import Libraries
import pandas as pd 
import numpy as np
```

```python
# Task 3: Data Cleaning – Dirty Cafe Sales Dataset

#Objective:
#Clean a messy dataset and prepare it for analysis using Python, Pandas, and NumPy.
```

```python
#Step 2: Load Dataset
df = pd.read_csv(r"C:\Users\ssima\Downloads\dirty_cafe_sales.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nData Types")
print(df.dtypes)
```

```python
#Standardized text formatting by converting item names to title case and converted transaction dates to datetime format.
```

```python
#change data type object to numeric
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Price Per Unit"] = pd.to_numeric(df["Price Per Unit"], errors="coerce")
df["Total Spent"] = pd.to_numeric(df["Total Spent"], errors="coerce")
df["Transaction Date"] = pd.to_datetime(
    df["Transaction Date"],
    errors="coerce"
)
```

```python
print("\nData Types")
print(df.dtypes)
```

```python
print("\nMissing Values")
print(df.isnull().sum())
```

```python
### Missing Value Handling

#- Item: Filled using mode because it is a categorical variable.
#- Quantity: Filled using median because numeric data may contain outliers.
#- Payment Method: Filled using mode.
#- Transaction Date: Filled using forward fill to preserve chronological order.
```

```python
df["Item"] = df["Item"].fillna(df["Item"].mode()[0])
df["Payment Method"] = df["Payment Method"].fillna(df["Payment Method"].mode()[0])
df["Location"] = df["Location"].fillna(df["Location"].mode()[0])
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Price Per Unit"] = df["Price Per Unit"].fillna(df["Price Per Unit"].median())
df["Total Spent"] = df["Total Spent"].fillna(
    df["Quantity"] * df["Price Per Unit"]
)
df["Total Spent"] = df["Total Spent"].fillna(df["Total Spent"].median())
df["Transaction Date"] = df["Transaction Date"].ffill()
```

```python
print("\nMissing Values")
print(df.isnull().sum())
```

```python
print("\nDuplicate Rows")
print(df.duplicated().sum())
```

```python
print("\nSummary Statistics")
print(df.describe(include='all'))
```
